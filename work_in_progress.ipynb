### Elméleti evezető

Az MLIP-ek célja, hogy közelítsék a kvantumkémiai, azaz a Schrödinger-egyenlet numerikus közelítő megoldásán alapuló *ab initio* számítások eredményeit. Az *ab initio* módszerek kémiai szempontból releváns prediktív képességét számos szakirodalmi példa igazolja. Alkalmazásuk ma már meghatározó szerepet tölt be többek között a kémiai reakciók mechanizmusának feltárásában,[1] a fizikai-kémiai folyamatok – például a termodinamikai viszonyok – értelmezésében,[2] az időfüggő jelenségek, például az elektronikus gerjesztések és optikai tulajdonságok leírásában,[3] valamint a spektroszkópiai jellemzők, például az IR- és NMR-spektrumok előrejelzésében.[4]

Az *ab initio* módszerek részletes bemutatása nem célja ennek a rövid jegyzetnek, mivel ezekről számos átfogó szakirodalmi forrás áll rendelkezésre [5]. Általánosságban a módszerek megkülönböztethetők az alkalmazott hullámfüggvény-ansatz, illetve a számítási pontosság és költség alapján. Utóbbi szempontból elkülöníthetők a kisebb számításigényű, ugyanakkor kevésbé pontos, valamint a nagyobb számítási költség mellett pontosabb eredményt szolgáltató eljárások.

Az MLIP-ek elméleti hátterének szempontjából különösen fontos a Born–Oppenheimer-közelítés, amely lehetővé teszi az elektronok és az atommagok mozgásának eltérő időskálán történő kezelését. Ennek értelmében az elektronikus Schrödinger-egyenletet rögzített atommag-konfiguráció mellett oldjuk meg:

$$
\hat{H}_{\mathrm{el}}(\mathbf{R})\Psi_{\mathrm{el}}
=
E_{\mathrm{el}}(\mathbf{R})\Psi_{\mathrm{el}},
$$

ahol $\mathbf{R}=(\mathbf{R}_1,\ldots,\mathbf{R}_N)$ az atommagok koordinátáit tartalmazó konfigurációs vektor, $\hat{H}_{\mathrm{el}}$ az elektronikus Hamilton-operátor, $\Psi_{\mathrm{el}}$ az elektronikus hullámfüggvény, míg $E_{\mathrm{el}}(\mathbf{R})$ az adott atommag-geometriához tartozó elektronikus energia.

A Born–Oppenheimer-közelítés keretében az atommagok mozgása egy effektív potenciálisenergia-felületen (*Potential Energy Surface*, PES) történik. A teljes potenciális energia az elektronikus energián túl az atommagok közötti Coulomb-kölcsönhatást is tartalmazza. Az MLIP feladata ennek a geometriafüggő energiafüggvénynek a közelítése:

$$
E_{\mathrm{ML}}(\mathbf{R})
\approx
E_{\mathrm{ref}}(\mathbf{R}),
$$

ahol $E_{\mathrm{ref}}$ a választott kvantumkémiai referenciaeljárással meghatározott energia, míg $E_{\mathrm{ML}}$ a betanított gépi tanulási modell predikciója.

Az energián túl az atomokra ható erők is meghatározó szerepet játszanak a molekuladinamikai szimulációkban. Az erők az energiafüggvény koordináták szerinti negatív gradienseként határozhatók meg:

$$
\mathbf{F}_i(\mathbf{R})
=
-\nabla_{\mathbf{R}_i}E(\mathbf{R}),
$$

ahol $\mathbf{F}_i$ az $i$-edik atomra ható erővektor.

Ennek megfelelően egy MLIP betanítása során nemcsak az energiaértékek, hanem az erők reprodukálása is optimalizálható. Egy általános, energia- és erőadatokat egyaránt figyelembe vevő veszteségfüggvény például a következő alakban írható fel:

$$
\mathcal{L}(\boldsymbol{\theta})
=
w_E
\sum_{k=1}^{N_{\mathrm{conf}}}
\left(
E_{\mathrm{ML}}^{(k)}-E_{\mathrm{ref}}^{(k)}
\right)^2
+
w_F
\sum_{k=1}^{N_{\mathrm{conf}}}
\sum_{i=1}^{N_k}
\sum_{\alpha\in\{x,y,z\}}
\left(
F_{\mathrm{ML},i\alpha}^{(k)}
-
F_{\mathrm{ref},i\alpha}^{(k)}
\right)^2,
$$

ahol $\boldsymbol{\theta}$ a modell tanulható paramétereit, $w_E$ és $w_F$ pedig az energia- és erőkomponensek relatív súlyát jelöli. $N_{\mathrm{conf}}$ a konfigurációk száma, $N_k$ pedig a $k$-adik konfigurációban található atomok száma.

Az MLIP-ek egyik legfontosabb gyakorlati előnye, hogy megfelelően összeállított referencia-adathalmaz esetén a nagy számításigényű kvantumkémiai módszerek eredményei lényegesen kisebb számítási költséggel reprodukálhatók. Ez azonban nem jelenti azt, hogy a modell általánosan helyettesítheti a referenciaeljárást. A predikció megbízhatósága alapvetően függ a tanítóadatok által lefedett konfigurációs tér kiterjedésétől, a modell reprezentációs képességétől, valamint a tanítás során alkalmazott optimalizációs eljárástól. A modell előrejelzései elsősorban azokra a geometriai konfigurációkra és kémiai környezetekre tekinthetők megbízhatónak, amelyek kellő reprezentációval rendelkeznek a tanítóadatokban.

Az adatkészletben az atomi rendszerek összenergiájából adódik. Ez a Born-Oppenheimer közelítésben az atomok saját energiáira + a kölcsönhatás energiáira bontható.

$$
E_{\mathrm{total}} =
\sum_{i=1}^{N} E_{\mathrm{atom}}(Z_i)
+ E_{\mathrm{interaction}}
$$

Ahhoz, hogy a modell jobb kezdeti paraméterekkel tanítható legyen, ezt célszerű levonni. Mivel az atomok saját energiái óriásiak a kémiailag releváns kölcsönhatás energiához képest, így stabilabb kezdeti paramétereket kapunk. Ezt végi a *RemoveOffset*

$$
E_{\mathrm{residual}} =
E_{\mathrm{total}} -
\sum_Z n_Z E_{\mathrm{ref}}(Z)
$$

Ezt a végén vissza kell állítani, így a neurális háló által viszaadott energia értékhez hozzá kell majd adni az atomok saját energiáit.
$$
E_{\mathrm{pred}} =
E_{\mathrm{NN}} +
\sum_Z n_Z E_{\mathrm{ref}}(Z)
$$

```python
cutoff = 5.0
n_atom_basis = 30

pairwise_distance = (
    spk.atomistic.PairwiseDistances()
)  # calculates pairwise distances between atoms
radial_basis = spk.nn.GaussianRBF(n_rbf=20, cutoff=cutoff)
schnet = spk.representation.SchNet(
    n_atom_basis=n_atom_basis,
    n_interactions=3,
    radial_basis=radial_basis,
    cutoff_fn=spk.nn.CosineCutoff(cutoff),
)
pred_U0 = spk.atomistic.Atomwise(n_in=n_atom_basis, output_key=QM9.U0)

nnpot = spk.model.NeuralNetworkPotential(
    representation=schnet,
    input_modules=[pairwise_distance],
    output_modules=[pred_U0],
    postprocessors=[
        trn.CastTo64(),
        trn.AddOffsets(QM9.U0, add_mean=True, add_atomrefs=True),
    ],
)
```
